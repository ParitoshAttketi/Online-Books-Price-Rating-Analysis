```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df= pd.read_csv("C:\\Users\\bhoom\\OneDrive\\Desktop\\book_scraping\\Books_online_cleaned_dataset.csv")
df
```

```python
df.columns
```

```python
plt.hist(df["Price in £"], bins=20)
plt.xlabel("Price in £")
plt.ylabel("Number of Books")
plt.title("Price in £ Distribution")
plt.show()
```

```python
plt.hist(df["Rating"], bins=10)
plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Rating Distribution")
plt.show()
```

```python
df["Category"].value_counts().head().plot(kind="bar")

plt.xlabel("Category")
plt.ylabel("Number of Books")
plt.title("Books by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.scatter(df["Rating"], df["Price in £"])

plt.xlabel("Rating")
plt.ylabel("Price in £")
plt.title("Rating vs Price in £")
plt.show()
```

```python
df["Availability"].value_counts().plot(kind="bar")

plt.xlabel("Availability")
plt.ylabel("Number of Books")
plt.title("Book Availability")
plt.show()
```

```python
top10 = df.sort_values("Number of Reviews", ascending=False).head(10)

plt.barh(top10["Name"], top10["Number of Reviews"])

plt.xlabel("Number of Reviews")
plt.ylabel("Book Name")
plt.title("Top 10 Books by Reviews")
plt.show()
```

```python
avg_price = df.groupby("Category")["Price in £"].mean().sort_values(ascending=False).head(10)

avg_price.plot(kind="bar")

plt.xlabel("Category")
plt.ylabel("Average Price")
plt.title("Average Price by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
avg_rating = df.groupby("Category")["Rating"].mean()

avg_rating.plot(kind="bar")

plt.xlabel("Category")
plt.ylabel("Average Rating")
plt.title("Average Rating by Category")
plt.xticks(rotation=90)
plt.show()
```

```python
avg_price = df.groupby("Category")["Price in £"].mean().sort_values(ascending=False)

print(avg_price)
```

```python
avg_price.plot(kind="bar", figsize=(10, 5))

plt.xlabel("Category")
plt.ylabel("Average Price")
plt.title("Average Price by Category")
plt.xticks(rotation=90)
plt.show()
```

```python
category_counts = df["Category"].value_counts()

print(category_counts)
```

```python
category_counts.plot(kind="bar", figsize=(10, 5))

plt.xlabel("Category")
plt.ylabel("Number of Books")
plt.title("Number of Books by Category")
plt.xticks(rotation=90)
plt.show()
```

```python
available = (df["Availability"] == "In stock").sum()

percentage = (available / len(df)) * 100

print("Percentage of books available:", percentage, "%")
```

```python
df["Availability"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Book Availability")
plt.show()
```
